```python
missing = df.isnull().sum()

missing = missing[missing > 0]

print(missing)
```

```python
print("Before:", len(df))

df = df.drop_duplicates()

print("After:", len(df))
```

```python
print(df.dtypes)
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
print(df[["Order Date", "Ship Date"]].dtypes)
```

```python
df[["Sales", "Quantity", "Discount", "Profit"]].describe()
```

```python
columns = ["Sales", "Quantity", "Discount", "Profit"]

for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    
    print(col, "outliers:", len(outliers))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.boxplot(data=df[["Sales", "Quantity", "Discount", "Profit"]])

plt.title("Boxplot of Numerical Variables")
plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
```

```python
plt.figure(figsize=(8, 5))

category_sales.plot(kind="bar")

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=0)

plt.show()
```

```python
category_profit = df.groupby("Category")["Profit"].sum()

print(category_profit)
```

```python
plt.figure(figsize=(8, 5))

category_profit.plot(kind="bar")

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.xticks(rotation=0)

plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

print(region_sales)
```

```python
plt.figure(figsize=(8, 5))

region_sales.plot(kind="bar")

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.xticks(rotation=0)

plt.show()
```

```python

df["Month"] = df["Order Date"].dt.to_period("M")
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()

print(monthly_sales)
```

```python
plt.figure(figsize=(12, 5))

monthly_sales.plot(kind="line")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)

plt.show()
```

```python
product_sales = df.groupby("Product Name")["Sales"].sum()
```

```python
top_10_products = product_sales.sort_values(ascending=False).head(10)

print(top_10_products)
```

```python
plt.figure(figsize=(10, 6))

top_10_products.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(df["Sales"], df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python
correlation = df["Sales"].corr(df["Profit"])

print("Correlation between Sales and Profit:", correlation)
```

```python
subcategory_sales = df.groupby("Sub-Category")["Sales"].sum()

print(subcategory_sales.sort_values(ascending=False))
```

```python
plt.figure(figsize=(10, 6))

subcategory_sales.sort_values().plot(kind="barh")

plt.title("Sales by Sub-Category")
plt.xlabel("Total Sales")
plt.ylabel("Sub-Category")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(df["Discount"], df["Profit"])

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()
```

```python
print("Discount-Profit correlation:",
      df["Discount"].corr(df["Profit"]))
```

```python
plt.figure(figsize=(8, 6))

correlation = df[["Sales", "Quantity", "Discount", "Profit"]].corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```
